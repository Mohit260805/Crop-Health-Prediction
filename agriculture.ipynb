```python
Q1 = data['Expected_Yield'].quantile(0.25)
Q3 = data['Expected_Yield'].quantile(0.75)

IQR = Q3-Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

data = data[(data['Expected_Yield']>=lower) & (data['Expected_Yield']>=lower)]
```

```python
data.shape
```

# SPLIT THE DATA INTO INPUT AND OUTPUT

```python
x = data.iloc[:,:-1]
x.drop(columns=['Expected_Yield','Ground_Truth_Segmentation'], inplace=True)
y = data[['Crop_Health_Label']]
```

# TRAIN THE MODEL

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_train = scaler.fit_transform(x_train)
```

```python
x_test = scaler.transform(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

## XGBOOST CLASSIFIER

```python
from xgboost import  XGBClassifier
```

```python
model = XGBClassifier(
    n_estimators=100, 
    random_state=42
)
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

### ACCURACY OF THE MODEL

```python
accuracy_score(y_test, y_pred)
```

```python
import joblib
```

```python
joblib.dump(model,'model.pkl')
```

## Logistic Regression

```python
y_pred = clf.predict(x_test)
```
